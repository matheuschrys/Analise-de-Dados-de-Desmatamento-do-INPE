# Análise de Dados de Desmatamento do INPE (PRODES)

Notebook reorganizado para usar caminhos relativos e funções reutilizáveis do pacote `src/desmatamento_inpe`.

A análise cruza a base municipal do PRODES com metadados de municípios brasileiros e gera agregações por ano, bioma e estado.

## 1. Configuração do ambiente

A célula abaixo localiza a raiz do projeto, habilita a importação do pacote local e carrega as bibliotecas necessárias.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
SRC_DIR = PROJECT_ROOT / "src"

if str(SRC_DIR) not in sys.path:
    sys.path.append(str(SRC_DIR))

import pandas as pd

from desmatamento_inpe import (
    DEFAULT_MUNICIPIOS_PATH,
    DEFAULT_PRODES_PATH,
    carregar_dados,
    calcular_area_total_por_bioma,
    calcular_desmatamento_por_bioma,
    calcular_percentual_desmatado_por_bioma,
    calcular_ranking_estados,
)
from desmatamento_inpe.visualization import (
    configurar_tema,
    plotar_area_total_monitorada,
    plotar_evolucao_desmatamento,
    plotar_percentual_desmatado,
    plotar_ranking_absoluto_estados,
    plotar_ranking_relativo_estados,
)

pd.options.display.float_format = "{:.2f}".format
configurar_tema()
```

## 2. Carga e integração dos dados

Os arquivos CSV ficam em `data/raw`, evitando caminhos absolutos dependentes da máquina do autor.

```python
print(DEFAULT_PRODES_PATH.relative_to(PROJECT_ROOT))
print(DEFAULT_MUNICIPIOS_PATH.relative_to(PROJECT_ROOT))
```

```python
df_completo = carregar_dados()
df_completo.head()
```

```python
df_completo.info()
```

```python
df_completo.describe()
```

## 3. Evolução do desmatamento por bioma

Agregação anual da área desmatada por bioma, com conversão para milhões de km² para facilitar a leitura dos gráficos.

```python
df_desmatamento_bioma = calcular_desmatamento_por_bioma(df_completo)
df_desmatamento_bioma.head()
```

```python
plotar_evolucao_desmatamento(df_desmatamento_bioma);
```

## 4. Área total monitorada por bioma

Essa visão ajuda a comparar o tamanho da área monitorada em cada bioma ao longo dos anos.

```python
df_area_total_bioma = calcular_area_total_por_bioma(df_completo)
df_area_total_bioma.head()
```

```python
plotar_area_total_monitorada(df_area_total_bioma);
```

## 5. Percentual de área desmatada por bioma

A métrica relativa mostra a proporção da área total que aparece como desmatada em cada bioma e ano.

```python
df_percentual_bioma = calcular_percentual_desmatado_por_bioma(df_completo)
df_percentual_bioma.head()
```

```python
plotar_percentual_desmatado(df_percentual_bioma);
```

## 6. Rankings por estado

Os rankings abaixo comparam os estados por volume absoluto de desmatamento e por percentual em relação à área total monitorada.

```python
df_estados = calcular_ranking_estados(df_completo)
df_estados.head()
```

```python
plotar_ranking_absoluto_estados(df_estados);
```

```python
plotar_ranking_relativo_estados(df_estados);
```
